## Задание

Необходимо вычислить значение выражения:
$$F=a\_1+b-a\cdot(x+y5)$$
Для каждой переменной необходимо вывести ее значение и тип.


## Анализ выражения и определение переменных

Проанализируем выражение и определим переменные, которые потребуются для выполнения задания. Независимые переменные: $x, y5, b, a, a\_1$. Переменная, которая должна быть вычислена на основе независимых: $F$.

```python
x=7
y5=17
b=5
a=8
a_1=4
print("x = ", x, ", type(x) = ", type(x))
print("y5 = ", y5, ", type(y5) = ", type(y5))
print("b = ", b, ", type(b) = ", type(b))
print("a = ", a, ", type(a) = ", type(a))
print("a_1= ", a_1, ", type(a_1) = ", type(a_1))
```

## Вычисление выражение, печать результатов

Код для вычисления значения выражения $F$:

```python
F=a_1+b-a*(x+y5)
print('F=',F)
print('Type of F: ', type(F))
```

Основные функции, используемые в данной практической работе: `print(), type(), abs()`
