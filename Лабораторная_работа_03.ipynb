##Ответы на контрольные вопросы

### 1.Чем отличается функция с возвратом значения от функций без возврата значения?

Функция **с возвратом значения** (function) исполняет инструкции и возвращает результат обратно в ту часть программы, которая её вызвала. Для этого используется инструкция `return`. Возвращённое значение можно присвоить переменной, вывести на экран или использовать в выражениях. Примеры: `input()`, `int()`, `math.sqrt()`.

Функция **без возврата значения** (процедура, procedure) просто исполняет содержащиеся в ней инструкции и завершается, не возвращая результат. Она выполняет некоторое действие (например, вывод сообщения на экран), но не передаёт значение обратно. Пример: функция `print()` ничего не возвращает (возвращает `None`).

Главное отличие: наличие или отсутствие инструкции `return` с выражением, возвращающим значение.


### 2.Что такое библиотечная функция?

**Библиотечная функция** — это функция, которая входит в стандартную библиотеку Python и уже написана разработчиками языка. Эти функции упрощают работу программиста, так как с их помощью решаются многие типовые задачи без необходимости писать код вручную.

Некоторые библиотечные функции встроены прямо в интерпретатор (например, `print`, `input`, `range`, `len`), и их можно вызывать напрямую. Другие хранятся в модулях (например, `math`, `random`) и требуют импорта с помощью инструкции `import`.

### 3.Почему библиотечные функции похожи на «чёрные ящики»?

Библиотечные функции рассматриваются как «чёрные ящики», потому что программист использует их, **не зная внутреннего устройства**. Термин «чёрный ящик» описывает механизм, который принимает нечто на входе, выполняет с ним некоторую работу (которую невозможно наблюдать) и выдаёт результат на выходе.

Программисту достаточно знать: какие аргументы передать функции и какой результат она вернёт. Как именно функция выполняет вычисления — скрыто внутри её реализации. Это позволяет абстрагироваться от деталей и сосредоточиться на решении задачи.

### 4.Что делает приведенная ниже инструкция?
```
x = random.randint(1, 100)
```

Эта инструкция генерирует случайное целое число в диапазоне от 1 до 100 (включительно) с помощью функции `randint` из модуля `random` и присваивает его переменной `x`.

### 5.Что делает приведенная ниже инструкция?
```
print(random.randint{1, 20))
```

Эта инструкция содержит **синтаксическую ошибку**: используются фигурные скобки `{` вместо круглых скобок `(` при вызове функции `randint`, а также несогласованные скобки (одна фигурная и одна круглая в конце).

Если исправить ошибку:
```python
print(random.randint(1, 20))
```
то инструкция сгенерирует случайное целое число от 1 до 20 (включительно) и выведет его на экран.

### 6.Для чего используется следующий вызов библиотечной функции?
```
random.seed(20)
```

Функция `random.seed(20)` инициализирует генератор случайных чисел начальным значением (seed), равным 20. Это делает последовательность «случайных» чисел **воспроизводимой**: при каждом запуске программы с одним и тем же значением seed генератор будет выдавать одну и ту же последовательность чисел.

Это полезно для отладки, тестирования и научных экспериментов, где требуется повторяемость результатов.

### 7.Каким образом функции помогают повторно использовать код в программе?

Функции позволяют написать код один раз и вызывать его многократно из разных мест программы. Вместо дублирования одинаковых инструкций программист определяет функцию, а затем вызывает её по имени wherever нужно. Это:

- **сокращает объём кода** и упрощает его поддержку;
- **упрощает командную работу** — разные программисты могут писать разные функции;
- **повышает надёжность** — исправление ошибки в функции автоматически исправляет её во всех местах вызова;
- **улучшает читаемость** — код разбивается на логические блоки с понятными именами.

Кроме того, функции можно переносить между программами, помещая их в модули.

### 8.Назовите и опишите две части определения функции.

1. **Заголовок функции** — первая строка определения. Начинается с ключевого слова `def`, за которым следует имя функции, круглые скобки с параметрами (если они есть) и двоеточие. Пример: `def distance(x, y):`

2. **Блок (тело) функции** — набор инструкций, составляющих функцию. Все инструкции выделены отступом. Блок исполняется при каждом вызове функции. В функции с возвратом значения в блоке обязательно присутствует инструкция `return`.

### 9.Как у функции задать параметры по умолчанию?

Параметры по умолчанию задаются в заголовке функции: параметру присваивается значение прямо в определении. Если при вызове функции аргумент для такого параметра не передан, используется значение по умолчанию.

```python
def distance(x, y=0):
    return math.sqrt(x ** 2 + y ** 2)
```

Теперь `distance(5)` вычислит расстояние до точки (5, 0), а `distance(3, 4)` — до точки (3, 4).

Параметры по умолчанию должны располагаться **после** параметров без значений по умолчанию.

````python
# Демонстрация параметров по умолчанию
def distance_default(x, y=0):
    """Расстояние от начала координат, y по умолчанию = 0."""
    return math.sqrt(x ** 2 + y ** 2)

print(f"distance_default(5)    = {distance_default(5)}")       # y = 0 по умолчанию
print(f"distance_default(3, 4) = {distance_default(3, 4)}")   # оба аргумента переданы
````

### 10.Что такое рекурсия?

**Рекурсия** — это приём программирования, при котором функция вызывает сама себя. Такая функция называется **рекурсивной**.

Рекурсивная функция обычно содержит:
1. **Базовый случай** (условие выхода) — когда функция перестаёт вызывать себя и возвращает результат.
2. **Рекурсивный случай** — когда функция вызывает саму себя с изменёнными аргументами.

Пример — вычисление суммы чисел от 1 до N:

```python
def sum_n(n):
    if n == 1:
        return 1
    return n + sum_n(n - 1)
```

Любую рекурсию можно представить в виде цикла (`for` или `while`), и наоборот. Рекурсия часто делает код более выразительным для задач, имеющих рекурсивную структуру (факториал и др.).

````python
# Демонстрация рекурсии: факториал
def factorial(n):
    """Рекурсивное вычисление факториала."""
    if n <= 1:          # базовый случай
        return 1
    return n * factorial(n - 1)  # рекурсивный случай

print(f"5! = {factorial(5)}")
print(f"10! = {factorial(10)}")
````
