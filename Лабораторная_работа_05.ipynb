## Задание

В файле pr5_1.txt содержится последовательность целых чисел.
Элементы последовательности могут принимать целые значения от
0 до 10 000 включительно. Определите количество пар, в которых
оба элемента меньше, чем среднее арифметическое всех чисел в
файле, а их сумма оканчивается на 19. В ответе запишите два числа:
сначала количество найденных пар, а затем – минимальную сумму
элементов таких пар. В данной задаче под парой подразумевается
два идущих подряд элемента последовательности.

## Решение

Файл pr5.txt расположен репозитории. Выполним копирование файла в локальное хранилище Google Colab, используя следующую команду:

````python
!wget https://raw.githubusercontent.com/enikolaev/IT-and-Programming/refs/heads/main/data/pract5/pr5_1.txt
````

Можно скопировать и поместить файл с данными любым удобным способом

Реализуем чтение строк из файла и конвертацию отдельных строк в числовые значения. Каждое получаемое число будем добавлять в список `data`.

````python
# Создаем пустой список
data = []
# Выполняем операции чтения и заполнение списка
with open('pr5_1.txt') as F:
  for line in F:
    digit = int(line)
    data.append(digit)

# Посмотрим, что мы считали...
print(len(data))
print(data[:10])
````

Теперь ответим на поставленные вопросы использую список `data`

````python
# Вычислим среднее арифметическое всех чисел в файле
sr = sum(data) / len(data)
print(sr)
````

````python
# Подсчет подходящих пар
counter = 0
min_sum = 20001  # больше любой возможной суммы (обе величины не более 10000)
for i in range(len(data) - 1):
  # пара соседних элементов
  x, y = data[i], data[i+1]
  # проверяем условия варианта
  if x < sr and y < sr and str(x + y)[-2:] == '19':
    counter += 1
    min_sum = min(min_sum, x + y)

print(f'Количество подходящих пар = {counter}')
print(f'Минимальная сумма = {min_sum}')
````

**4. Какую инструкцию следует применить для добавления в список `names` в индексную позицию 0 строкового значения 'Вэнди'? Почему именно эту инструкцию вместо другой?**

Правильный вариант — **а)** `names[0] = 'Вэнди'` (с квадратными скобками).

Пункт «а» — это присваивание значения элементу по индексу: элемент с индексом 0 получает строку 'Вэнди', то есть значение попадает ровно в заданную индексную позицию 0. Вариант «б» `names.append('Вэнди')` тоже изменит список, но метод `append()` всегда дописывает элемент **в конец** списка, а не в позицию 0, поэтому он не соответствует условию задачи. (Дополнительный корректный способ вставить значение в начало списка — метод `insert()`: `names.insert(0, 'Вэнди')`.)

**5. Опишите приведенные ниже списковые методы:**

- **а) `index()`** — ищет первое вхождение значения в список и возвращает его индекс. Например, `numbers.index(3)` вернет позицию, на которой впервые встречается число 3. Если значения в списке нет, возбуждается исключение `ValueError`.
- **б) `insert()`** — вставляет элемент в список по указанному индексу, сдвигая все элементы правее на одну позицию вправо: `numbers.insert(0, 'Вэнди')` вставит строку в начало списка.
- **в) `sort()`** — сортирует список на месте, то есть изменяет сам список, не создавая новый. По умолчанию сортировка выполняется по возрастанию (`numbers.sort()`); для сортировки по убыванию используется параметр: `numbers.sort(reverse=True)`.
- **г) `reverse()`** — переставляет элементы списка в обратном порядке (последний элемент становится первым и так далее). Как и `sort()`, метод работает «на месте» и ничего не возвращает.
