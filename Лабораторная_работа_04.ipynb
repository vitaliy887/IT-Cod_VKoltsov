## Задание

В файле data_v1.txt содержится последовательность символов, содержащая буквы латинского алфавита и десятичные цифры. Необходимо решить следующие задачи:
В файле data_v2.txt посчитайте количество цифр, расположенных
рядом с символом «Z»


## Анализ задания, получение данных

Файл data_v1.txt расположен репозитории. Выполним копирование файла в локальное хранилище Google Colab, используя следующую команду:

````python
!wget https://raw.githubusercontent.com/enikolaev/IT-and-Programming/main/data/data_v1.txt
````

Можно скопировать и поместить файл с данными любым удобным способом

##Решение задания:

````python
f=open('data_v1.txt')
s=f.read().strip()
f.close()
print(f"Длина строки: {len(s)}")

# Шаг 3. Подсчёт цифр, расположенных рядом с символом «Z»
count = 0
for i in range(len(s)):
    if s[i] == 'Z':
        # Проверяем символ слева от Z
        if i > 0 and s[i - 1].isdigit():
            count += 1
        # Проверяем символ справа от Z
        if i < len(s) - 1 and s[i + 1].isdigit():
            count += 1

print(f"Количество цифр, расположенных рядом с символом 'Z': {count}")
````

###Количество цифр, расположенных рядом с символом «Z» = 539


## Ответы на контрольные вопросы:


###1.Назначение, сходства и различия `read()`, `readline()`, `readlines()`

Все три метода служат для чтения данных из файла, открытого в режиме `'r'`. Различия:

`read()` — считывает всё содержимое файла целиком и возвращает его одной строкой. Удобен для небольших файлов.

`readline()` — считывает одну строку файла (до символа `\n` включительно). При каждом следующем вызове продолжает чтение со следующей строки. Удобен для построчной обработки.

`readlines()` — считывает все строки файла и возвращает их списком, где каждый элемент — одна строка. Удобен, когда нужен случайный доступ к строкам по индексу.

Сходство: все три метода работают с файловым объектом и не требуют явного указания количества строк (кроме `read()`, которому можно передать количество символов).

###2.Как прочитать многострочный файл, не используя `read()`, `readline()`, `readlines()`?

В Python файловый объект является итерируемым — по нему можно пройти циклом `for` напрямую:

```python
f = open('data.txt')
for line in f:
    print(line.strip())
f.close()
```
На каждой итерации переменная `line` получает очередную строку файла. Это особенно удобно для больших файлов, так как строки загружаются в память по одной, а не все сразу.

###3.Задача закрытия файла

Закрытие файла (метод `close()`) разрывает связь между файловым объектом и файлом на диске. Это необходимо, чтобы:

освободить системные ресурсы, занятые открытым файлом;

*   освободить системные ресурсы, занятые открытым файлом;

*   гарантировать, что все буферизованные данные действительно записаны на диск (особенно при записи);


*   избежать блокировки файла другими программами.

Если файл не закрыть, данные могут остаться в буфере и не попасть на диск.

###4.Позиция считывания файла

Позиция считывания — это внутренний указатель файлового объекта, который отмечает, с какого места в файле будет прочитан следующий фрагмент данных. При открытии файла для чтения (режим `'r'`) позиция считывания находится в **начале файла** — на нулевом байте. После каждого вызова `read()` или `readline()` позиция сдвигается вперёд на количество прочитанных символов.

###5.Режим записи без удаления существующего содержимого

Для дозаписи файл открывается в режиме `'a'` (от append). При этом существующее содержимое файла остаётся нетронутым, а новые данные **дописываются в конец файла.** Если файла с таким именем не существует, он будет создан.

###6.Что такое исключение

Исключение — это ошибка, которая происходит во время работы программы. В большинстве случаев исключение приводит к внезапному аварийному останову программы. Например, деление на ноль или попытка преобразовать нечисловую строку в число вызывают исключение.

###7.Что происходит, если исключение не обрабатывается $try/except$

Если исключение вызвано, а программа не обрабатывает его инструкцией $try/except$, программа аварийно останавливается и выводит сообщение об ошибке с отчётом обратной трассировки (traceback) — стек вызовов, показывающий, где именно произошла ошибка.

###8.Исключение при открытии несуществующего файла

При попытке открыть несуществующий файл для чтения вызывается исключение $FileNotFoundError$.

###9.Исключение при конвертации нечисловой строки через `float()`

При применении `float()` к строковому значению, которое не может быть интерпретировано как число, вызывается исключение $ValueError$.

###10.Открытие несуществующего файла в режиме дозаписи

Если файл не существует и программа пытается открыть его в режиме дозаписи `('a')`, **ошибки не возникает** — файл автоматически создаётся и открывается для записи. Если файл уже существует, данные дописываются в его конец без удаления прежнего содержимого.
